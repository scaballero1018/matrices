```python
%%writefile menu.py

def mostrar_menu():

    print("\n--- MENU DE OPERACIONES CON MATRICES ---")
    print("1. Suma de matrices")
    print("2. Multiplicacion de matrices")
    print("3. Producto de Hadamard")
    print("4. Producto de Kronecker")
    print("5. Salir (si desea terminar)")

    while True:
        try:
            opcion = int(input("Seleccione una opcion: "))

            if opcion in [1, 2, 3, 4, 5]:
                return opcion

            else:
                print("Opcion no valida")

        except ValueError:
            print("Error: debe ingresar un numero.")
```

```python
%%writefile entrada.py

def ingresar_matriz():

    while True:
        try:
            filas = int(input("Ingrese el numero de filas de la matriz: "))
            columnas = int(input("Ingrese el numero de columnas de la matriz: "))

            if filas <= 0 or columnas <= 0:
                print("Debe ingresar numeros enteros positivos.")
                continue

            matriz = []

            for i in range(filas):
                while True:
                    try:
                        print("Ingrese los valores de la fila", i + 1, "separados por espacios:")
                        fila = input().split()

                        if len(fila) != columnas:
                            print("Error: debe ingresar exactamente", columnas, "valores (recuerde la candidad de columnas).")
                            continue

                        fila_numeros = []

                        for valor in fila:
                            fila_numeros.append(float(valor))

                        matriz.append(fila_numeros)
                        break

                    except ValueError:
                        print("Error: debe ingresar solo numeros.")

            return matriz

        except ValueError:
            print("Error: debe ingresar numeros enteros.")


def mostrar_matriz(A):

    print("La matriz es:")

    for fila in A:
        print(fila)     #casi que no podia por la sangriaa
```

```python
%%writefile operaciones_matrices.py

def sumar_matrices(A, B):

    filas = len(A)
    columnas = len(A[0])

    if filas != len(B) or columnas != len(B[0]):
        print("Error: Las matrices deben tener las mismas dimensiones.")
        return None

    resultado = []

    for i in range(filas):
        fila = []
        for j in range(columnas):
            fila.append(A[i][j] + B[i][j])
        resultado.append(fila)

    return resultado


def multiplicar_matrices(A, B):

    filas_A = len(A)
    columnas_A = len(A[0])
    filas_B = len(B)
    columnas_B = len(B[0])

    if columnas_A != filas_B:
        print("Error: No se pueden multiplicar las matrices.")
        return None

    resultado = []

    for i in range(filas_A):
        fila = []
        for j in range(columnas_B):
            suma = 0
            for k in range(columnas_A):
                suma = suma + A[i][k] * B[k][j]
            fila.append(suma)
        resultado.append(fila)

    return resultado


def hadamard_matrices(A, B):

    filas = len(A)
    columnas = len(A[0])

    if filas != len(B) or columnas != len(B[0]):
        print("Error: Las matrices deben tener las mismas dimensiones.")
        return None

    resultado = []

    for i in range(filas):
        fila = []
        for j in range(columnas):
            fila.append(A[i][j] * B[i][j])
        resultado.append(fila)

    return resultado


def kronecker(A, B):

    filas_A = len(A)
    columnas_A = len(A[0])
    filas_B = len(B)
    columnas_B = len(B[0])

    resultado = []

    for i in range(filas_A):
        for k in range(filas_B):
            fila = []
            for j in range(columnas_A):
                for l in range(columnas_B):
                    fila.append(A[i][j] * B[k][l])
            resultado.append(fila)

    return resultado
    #debo admitir que aqui utilice ia era un poco distinto a lo que hbia hecho con
    # los complejos en el trabajo anterio y gran parte de lo he hecho hasta hora
    #me he guiado con ese trabajo
```

```python
%%writefile main.py

from menu import mostrar_menu
from entrada import ingresar_matriz
from entrada import mostrar_matriz
from operaciones_matrices import sumar_matrices
from operaciones_matrices import multiplicar_matrices
from operaciones_matrices import hadamard_matrices
from operaciones_matrices import kronecker


opcion = 0

while opcion != 5:

    opcion = mostrar_menu()

    if opcion == 1:
        print("Ingrese la primera matriz:")
        A = ingresar_matriz()

        print("Ingrese la segunda matriz (recuerde que deben tener las mismas dimensiones):")
        B = ingresar_matriz()

        resultado = sumar_matrices(A, B)

        if resultado is not None:
            print("Resultado de la suma:")
            mostrar_matriz(resultado)
        else:
            print("No se pueden sumar las matrices.")
            print("Las matrices deben tener el mismo número de filas y columnas.")


    elif opcion == 2:
        print("Ingrese la primera matriz:")
        A = ingresar_matriz()

        print("Ingrese la segunda matriz:")
        print("Recuerde que el # de columnas de la primera matriz debe ser igual al # de filas de la segunda matriz ")
        B = ingresar_matriz()

        resultado = multiplicar_matrices(A, B)

        if resultado is not None:
            print("Resultado de la multiplicacion:")
            mostrar_matriz(resultado)
        else:
            print("No se puede multiplicar las matrices.")
            print("El número de columnas de la primera matriz debe ser igual")
            print("al número de filas de la segunda matriz.")


    elif opcion == 3:
        print("Ingrese la primera matriz:")
        A = ingresar_matriz()

        print("Ingrese la segunda matriz:")
        print("Recuerde que las matrices deben tener las mismas dimensiones.")
        B = ingresar_matriz()

        resultado = hadamard_matrices(A, B)

        if resultado is not None:
            print("Resultado del producto de Hadamard:")
            mostrar_matriz(resultado)
        else:
            print("No se puede realizar el producto de Hadamard.")
            print("Las matrices deben tener las mismas dimensiones.")


    elif opcion == 4:
        print("Ingrese la primera matriz:")
        print("puedes multiplicar matrices de cualquier tamaño.")
        A = ingresar_matriz()

        print("Ingrese la segunda matriz:")
        B = ingresar_matriz()

        resultado = kronecker(A, B)

        print("Resultado del producto de Kronecker:")
        mostrar_matriz(resultado)


    elif opcion == 5:
        print("Gracias por usar el programa")
        print("Fin del programa")
```
